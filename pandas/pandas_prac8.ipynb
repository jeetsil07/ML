```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "city" : ["new york","chicago","orlando","baltimore"],
    "temperature": [21,14,35,32]
})
df1
```

```python
df2 = pd.DataFrame({
    "city" : ["chicago","new york","san fransisco"],
    "humidity": [65,68,71]
})
df2
```

```python
df3 = pd.merge(df1,df2, on = "city", how="outer", indicator = True)
df3
```

```python
df1 = pd.DataFrame({
    "city" : ["new york","chicago","orlando","baltimore"],
    "temperature": [21,14,35,32],
    "humidity": [65,68,71,75]
})
df1
```

```python
df2 = pd.DataFrame({
    "city" : ["chicago","new york","san diego"],
    "temperature": [21,14,35],
    "humidity": [65,68,71]
})
df2
```

```python
df3 = pd.merge(df1,df2, on = "city", how="outer", suffixes = '')
df3
```
